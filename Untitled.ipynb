```python
#!/usr/bin/env python3
"""
Jupyter Notebook Script for MAPF Results Visualization
Generates one plot per map showing performance across different agent counts
"""

import glob
import re
import os
from collections import defaultdict
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Configure matplotlib for better plots
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (16, 10)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.labelsize'] = 12
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['legend.fontsize'] = 10
plt.rcParams['xtick.labelsize'] = 10
plt.rcParams['ytick.labelsize'] = 10

# %matplotlib inline  # Uncomment this line when running in Jupyter

# ============================================================================
# DATA PARSING FUNCTIONS
# ============================================================================

def parse_progress_logs(progress_dir='batch_results/progress_logs'):
    """
    Parse all progress log files to extract metrics by map and agent count.
    
    Returns:
        dict[map_name][num_agents] = {
            'solved': count,
            'total': count,
            'failed': count,
            'timeouts': count,
            'runtimes': [list of floats],
            'makespans': [list of ints],
            'socs': [list of ints],
            'timeout_limits': [list of floats]
        }
    """
    data = defaultdict(lambda: defaultdict(lambda: {
        'solved': 0,
        'total': 0,
        'failed': 0,
        'timeouts': 0,
        'runtimes': [],
        'makespans': [],
        'socs': [],
        'timeout_limits': []
    }))
    
    if not os.path.exists(progress_dir):
        print(f"Warning: Directory {progress_dir} does not exist!")
        return data
    
    log_files = glob.glob(f'{progress_dir}/progress_*.log')
    
    if not log_files:
        print(f"Warning: No progress log files found in {progress_dir}")
        return data
    
    print(f"Found {len(log_files)} progress log files")
    
    for log_file in log_files:
        # Extract map name and agent count from filename
        # Format: progress_mapname_N50.log
        basename = os.path.basename(log_file)
        match = re.search(r'progress_(.+?)_N(\d+)\.log', basename)
        if not match:
            print(f"Warning: Could not parse filename {basename}")
            continue
        
        map_name = match.group(1)
        num_agents = int(match.group(2))
        
        with open(log_file, 'r') as f:
            for line in f:
                line = line.strip()
                
                # Parse result lines (start with status icon)
                if line.startswith('✅') or line.startswith('⏱️') or line.startswith('❌'):
                    data[map_name][num_agents]['total'] += 1
                    
                    # Extract timeout limit
                    limit_match = re.search(r'limit=(\d+)s', line)
                    if limit_match:
                        data[map_name][num_agents]['timeout_limits'].append(float(limit_match.group(1)))
                    
                    # Check if solved
                    if 'SOLVED' in line:
                        data[map_name][num_agents]['solved'] += 1
                        
                        # Extract runtime
                        runtime_match = re.search(r'runtime=([\d.]+)s', line)
                        if runtime_match:
                            data[map_name][num_agents]['runtimes'].append(float(runtime_match.group(1)))
                        
                        # Extract makespan
                        makespan_match = re.search(r'makespan=(\d+)', line)
                        if makespan_match:
                            data[map_name][num_agents]['makespans'].append(int(makespan_match.group(1)))
                        
                        # Extract sum of cost
                        soc_match = re.search(r'soc=(\d+)', line)
                        if soc_match:
                            data[map_name][num_agents]['socs'].append(int(soc_match.group(1)))
                    
                    # Check if timeout
                    elif 'TIMEOUT' in line or line.startswith('⏱️'):
                        data[map_name][num_agents]['timeouts'] += 1
                    
                    # Check if failed
                    elif 'FAILED' in line or line.startswith('❌'):
                        data[map_name][num_agents]['failed'] += 1
    
    return data


def extract_summary_statistics(map_data):
    """
    Extract summary statistics for a single map across all agent counts.
    
    Returns:
        DataFrame with statistics for each agent count
    """
    rows = []
    
    for num_agents in sorted(map_data.keys()):
        metrics = map_data[num_agents]
        
        row = {
            'agents': num_agents,
            'total': metrics['total'],
            'solved': metrics['solved'],
            'timeouts': metrics['timeouts'],
            'failed': metrics['failed'],
            'success_rate': (metrics['solved'] / metrics['total'] * 100) if metrics['total'] > 0 else 0,
            'timeout_rate': (metrics['timeouts'] / metrics['total'] * 100) if metrics['total'] > 0 else 0,
            'avg_runtime': np.mean(metrics['runtimes']) if metrics['runtimes'] else 0,
            'std_runtime': np.std(metrics['runtimes']) if metrics['runtimes'] else 0,
            'avg_makespan': np.mean(metrics['makespans']) if metrics['makespans'] else 0,
            'std_makespan': np.std(metrics['makespans']) if metrics['makespans'] else 0,
            'avg_soc': np.mean(metrics['socs']) if metrics['socs'] else 0,
            'std_soc': np.std(metrics['socs']) if metrics['socs'] else 0,
            'avg_timeout_limit': np.mean(metrics['timeout_limits']) if metrics['timeout_limits'] else 0
        }
        rows.append(row)
    
    return pd.DataFrame(rows)


# ============================================================================
# PLOTTING FUNCTIONS
# ============================================================================

def plot_single_map_comprehensive(map_name, map_data, save_path=None):
    """
    Create a comprehensive 2x3 plot for a single map showing all key metrics.
    
    Subplots:
    1. Success/Timeout/Failed rates
    2. Average runtime with error bars
    3. Runtime vs Timeout limit
    4. Average makespan with error bars
    5. Average sum of cost with error bars
    6. Instance counts (solved/timeout/failed)
    """
    df = extract_summary_statistics(map_data)
    
    if df.empty:
        print(f"No data for {map_name}")
        return
    
    # Create figure with 2x3 subplots
    fig = plt.figure(figsize=(18, 12))
    gs = fig.add_gridspec(2, 3, hspace=0.3, wspace=0.3)
    
    fig.suptitle(f'Performance Analysis: {map_name}', 
                 fontsize=18, fontweight='bold', y=0.98)
    
    # Color scheme
    colors = {
        'success': '#2ecc71',
        'timeout': '#e74c3c',
        'failed': '#95a5a6',
        'runtime': '#3498db',
        'makespan': '#9b59b6',
        'soc': '#e67e22'
    }
    
    # ========== Plot 1: Success/Timeout/Failed Rates ==========
    ax1 = fig.add_subplot(gs[0, 0])
    ax1.plot(df['agents'], df['success_rate'], 'o-', linewidth=2.5, 
             markersize=9, label='Success Rate', color=colors['success'])
    ax1.plot(df['agents'], df['timeout_rate'], 's--', linewidth=2, 
             markersize=7, label='Timeout Rate', color=colors['timeout'])
    
    # Add failed rate if there are any failures
    if df['failed'].sum() > 0:
        failed_rate = (df['failed'] / df['total'] * 100)
        ax1.plot(df['agents'], failed_rate, '^:', linewidth=2, 
                 markersize=7, label='Failed Rate', color=colors['failed'])
    
    ax1.set_xlabel('Number of Agents', fontsize=13, fontweight='bold')
    ax1.set_ylabel('Rate (%)', fontsize=13, fontweight='bold')
    ax1.set_title('(a) Success vs Timeout Rate', fontsize=14, fontweight='bold', pad=10)
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best', framealpha=0.9)
    ax1.set_ylim([-5, 105])
    ax1.axhline(y=50, color='gray', linestyle=':', alpha=0.5)
    
    # ========== Plot 2: Average Runtime with Error Bars ==========
    ax2 = fig.add_subplot(gs[0, 1])
    ax2.errorbar(df['agents'], df['avg_runtime'], yerr=df['std_runtime'],
                 fmt='o-', linewidth=2.5, markersize=9, capsize=5, capthick=2,
                 color=colors['runtime'], ecolor=colors['runtime'], alpha=0.7)
    ax2.set_xlabel('Number of Agents', fontsize=13, fontweight='bold')
    ax2.set_ylabel('Runtime (seconds)', fontsize=13, fontweight='bold')
    ax2.set_title('(b) Runtime Scaling (mean ± std)', fontsize=14, fontweight='bold', pad=10)
    ax2.grid(True, alpha=0.3, linestyle='--')
    
    # Add exponential trend line if data shows growth
    if len(df) > 3:
        z = np.polyfit(df['agents'], df['avg_runtime'], 2)
        p = np.poly1d(z)
        ax2.plot(df['agents'], p(df['agents']), '--', alpha=0.5, 
                 color='gray', linewidth=2, label='Trend')
        ax2.legend(loc='best')
    
    # ========== Plot 3: Runtime vs Timeout Limit ==========
    ax3 = fig.add_subplot(gs[0, 2])
    ax3.plot(df['agents'], df['avg_timeout_limit'], 's--', linewidth=2, 
             markersize=7, label='Timeout Limit', color='#34495e', alpha=0.6)
    ax3.plot(df['agents'], df['avg_runtime'], 'o-', linewidth=2.5, 
             markersize=9, label='Avg Runtime', color=colors['runtime'])
    
    # Fill area between runtime and timeout
    ax3.fill_between(df['agents'], df['avg_runtime'], df['avg_timeout_limit'], 
                     alpha=0.2, color=colors['success'], label='Safety Margin')
    
    ax3.set_xlabel('Number of Agents', fontsize=13, fontweight='bold')
    ax3.set_ylabel('Time (seconds)', fontsize=13, fontweight='bold')
    ax3.set_title('(c) Runtime vs Timeout Limit', fontsize=14, fontweight='bold', pad=10)
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.legend(loc='best', framealpha=0.9)
    
    # ========== Plot 4: Average Makespan with Error Bars ==========
    ax4 = fig.add_subplot(gs[1, 0])
    ax4.errorbar(df['agents'], df['avg_makespan'], yerr=df['std_makespan'],
                 fmt='o-', linewidth=2.5, markersize=9, capsize=5, capthick=2,
                 color=colors['makespan'], ecolor=colors['makespan'], alpha=0.7)
    ax4.set_xlabel('Number of Agents', fontsize=13, fontweight='bold')
    ax4.set_ylabel('Makespan (timesteps)', fontsize=13, fontweight='bold')
    ax4.set_title('(d) Makespan Scaling (mean ± std)', fontsize=14, fontweight='bold', pad=10)
    ax4.grid(True, alpha=0.3, linestyle='--')
    
    # ========== Plot 5: Average Sum of Cost with Error Bars ==========
    ax5 = fig.add_subplot(gs[1, 1])
    ax5.errorbar(df['agents'], df['avg_soc'], yerr=df['std_soc'],
                 fmt='o-', linewidth=2.5, markersize=9, capsize=5, capthick=2,
                 color=colors['soc'], ecolor=colors['soc'], alpha=0.7)
    ax5.set_xlabel('Number of Agents', fontsize=13, fontweight='bold')
    ax5.set_ylabel('Sum of Cost', fontsize=13, fontweight='bold')
    ax5.set_title('(e) Sum of Cost Scaling (mean ± std)', fontsize=14, fontweight='bold', pad=10)
    ax5.grid(True, alpha=0.3, linestyle='--')
    
    # ========== Plot 6: Stacked Bar Chart of Outcomes ==========
    ax6 = fig.add_subplot(gs[1, 2])
    width = (df['agents'].max() - df['agents'].min()) / len(df) * 0.6
    
    ax6.bar(df['agents'], df['solved'], width=width, label='Solved', 
            color=colors['success'], alpha=0.8)
    ax6.bar(df['agents'], df['timeouts'], width=width, bottom=df['solved'],
            label='Timeout', color=colors['timeout'], alpha=0.8)
    ax6.bar(df['agents'], df['failed'], width=width, 
            bottom=df['solved'] + df['timeouts'],
            label='Failed', color=colors['failed'], alpha=0.8)
    
    ax6.set_xlabel('Number of Agents', fontsize=13, fontweight='bold')
    ax6.set_ylabel('Number of Instances', fontsize=13, fontweight='bold')
    ax6.set_title('(f) Instance Outcomes', fontsize=14, fontweight='bold', pad=10)
    ax6.grid(True, alpha=0.3, linestyle='--', axis='y')
    ax6.legend(loc='best', framealpha=0.9)
    
    # Add total count on top of each bar
    for i, agents in enumerate(df['agents']):
        total = df.iloc[i]['total']
        ax6.text(agents, total + total*0.02, str(int(total)), 
                ha='center', va='bottom', fontweight='bold', fontsize=10)
    
    # Save if path provided
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight', facecolor='white')
        print(f"✅ Saved plot: {save_path}")
    
    plt.show()
    
    # ========== Print Summary Statistics ==========
    print(f"\n{'='*80}")
    print(f"📊 SUMMARY STATISTICS: {map_name}")
    print(f"{'='*80}")
    print(f"Agent range: {df['agents'].min()} - {df['agents'].max()}")
    print(f"Total experiments: {df['total'].sum()}")
    print(f"Overall success rate: {df['solved'].sum() / df['total'].sum() * 100:.2f}%")
    print(f"Overall timeout rate: {df['timeouts'].sum() / df['total'].sum() * 100:.2f}%")
    
    if df['failed'].sum() > 0:
        print(f"Overall failed rate: {df['failed'].sum() / df['total'].sum() * 100:.2f}%")
    
    print(f"\nRuntime statistics:")
    print(f"  Min avg runtime: {df['avg_runtime'].min():.2f}s (at {df.loc[df['avg_runtime'].idxmin(), 'agents']:.0f} agents)")
    print(f"  Max avg runtime: {df['avg_runtime'].max():.2f}s (at {df.loc[df['avg_runtime'].idxmax(), 'agents']:.0f} agents)")
    
    if df['avg_makespan'].max() > 0:
        print(f"\nMakespan statistics:")
        print(f"  Min avg makespan: {df['avg_makespan'].min():.1f} (at {df.loc[df['avg_makespan'].idxmin(), 'agents']:.0f} agents)")
        print(f"  Max avg makespan: {df['avg_makespan'].max():.1f} (at {df.loc[df['avg_makespan'].idxmax(), 'agents']:.0f} agents)")
    
    if df['avg_soc'].max() > 0:
        print(f"\nSum of cost statistics:")
        print(f"  Min avg SOC: {df['avg_soc'].min():.1f} (at {df.loc[df['avg_soc'].idxmin(), 'agents']:.0f} agents)")
        print(f"  Max avg SOC: {df['avg_soc'].max():.1f} (at {df.loc[df['avg_soc'].idxmax(), 'agents']:.0f} agents)")
    
    print(f"{'='*80}\n")
    
    return df


def plot_cross_map_comparison(data, agent_count=50, save_path=None):
    """
    Compare all maps at a specific agent count.
    Shows success rate and runtime across different maps.
    """
    map_names = []
    success_rates = []
    timeout_rates = []
    avg_runtimes = []
    total_instances = []
    
    for map_name in sorted(data.keys()):
        if agent_count in data[map_name]:
            metrics = data[map_name][agent_count]
            map_names.append(map_name)
            
            success_rate = (metrics['solved'] / metrics['total'] * 100) if metrics['total'] > 0 else 0
            timeout_rate = (metrics['timeouts'] / metrics['total'] * 100) if metrics['total'] > 0 else 0
            success_rates.append(success_rate)
            timeout_rates.append(timeout_rate)
            
            avg_runtime = np.mean(metrics['runtimes']) if metrics['runtimes'] else 0
            avg_runtimes.append(avg_runtime)
            
            total_instances.append(metrics['total'])
    
    if not map_names:
        print(f"No data found for {agent_count} agents")
        return
    
    # Create figure with 2 subplots
    fig, axes = plt.subplots(1, 2, figsize=(18, 7))
    fig.suptitle(f'Cross-Map Comparison at {agent_count} Agents', 
                 fontsize=18, fontweight='bold')
    
    # Plot 1: Success and Timeout Rates
    ax1 = axes[0]
    x_pos = np.arange(len(map_names))
    width = 0.35
    
    bars1 = ax1.bar(x_pos - width/2, success_rates, width, 
                    label='Success Rate', color='#2ecc71', alpha=0.8)
    bars2 = ax1.bar(x_pos + width/2, timeout_rates, width,
                    label='Timeout Rate', color='#e74c3c', alpha=0.8)
    
    ax1.set_xlabel('Map', fontsize=13, fontweight='bold')
    ax1.set_ylabel('Rate (%)', fontsize=13, fontweight='bold')
    ax1.set_title('Success vs Timeout Rate by Map', fontsize=14, fontweight='bold')
    ax1.set_xticks(x_pos)
    ax1.set_xticklabels(map_names, rotation=45, ha='right')
    ax1.legend(loc='best', framealpha=0.9)
    ax1.grid(True, alpha=0.3, axis='y', linestyle='--')
    ax1.set_ylim([0, 105])
    
    # Add value labels on bars
    for bars in [bars1, bars2]:
        for bar in bars:
            height = bar.get_height()
            ax1.text(bar.get_x() + bar.get_width()/2., height,
                    f'{height:.0f}%', ha='center', va='bottom', fontsize=9)
    
    # Plot 2: Average Runtimes
    ax2 = axes[1]
    bars3 = ax2.bar(x_pos, avg_runtimes, color='#3498db', alpha=0.8)
    
    ax2.set_xlabel('Map', fontsize=13, fontweight='bold')
    ax2.set_ylabel('Average Runtime (seconds)', fontsize=13, fontweight='bold')
    ax2.set_title('Runtime by Map', fontsize=14, fontweight='bold')
    ax2.set_xticks(x_pos)
    ax2.set_xticklabels(map_names, rotation=45, ha='right')
    ax2.grid(True, alpha=0.3, axis='y', linestyle='--')
    
    # Add value labels and instance counts
    for i, (bar, instances) in enumerate(zip(bars3, total_instances)):
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height,
                f'{height:.1f}s\n(n={instances})', 
                ha='center', va='bottom', fontsize=9)
    
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight', facecolor='white')
        print(f"✅ Saved cross-map comparison: {save_path}")
    
    plt.show()


# ============================================================================
# MAIN EXECUTION
# ============================================================================

def main():
    """
    Main function to parse data and generate all plots.
    """
    print("="*80)
    print("🎨 MAPF RESULTS VISUALIZATION")
    print("="*80)
    print()
    
    # Parse the data
    print("📂 Parsing progress logs...")
    data = parse_progress_logs()
    
    if not data:
        print("❌ No data found! Make sure batch_results/progress_logs/ exists and contains log files.")
        return
    
    print(f"✅ Successfully parsed data for {len(data)} maps\n")
    
    # Display available maps
    print("📋 Available maps:")
    for map_name in sorted(data.keys()):
        agent_counts = sorted(data[map_name].keys())
        total_experiments = sum(data[map_name][n]['total'] for n in agent_counts)
        print(f"  • {map_name:30s} | {len(agent_counts):2d} agent configs | "
              f"{min(agent_counts):3d}-{max(agent_counts):3d} agents | "
              f"{total_experiments:3d} experiments")
    print()
    
    # Create output directory
    output_dir = 'batch_results/plots'
    os.makedirs(output_dir, exist_ok=True)
    print(f"📁 Plots will be saved to: {output_dir}/\n")
    
    # Generate plots for all maps
    print("="*80)
    print("🎨 GENERATING INDIVIDUAL MAP PLOTS")
    print("="*80)
    print()
    
    all_dataframes = {}
    
    for i, map_name in enumerate(sorted(data.keys()), 1):
        print(f"[{i}/{len(data)}] Generating plot for: {map_name}")
        save_path = f"{output_dir}/{map_name}_analysis.png"
        df = plot_single_map_comprehensive(map_name, data[map_name], save_path=save_path)
        all_dataframes[map_name] = df
        print()
    
    # Generate cross-map comparison
    print("="*80)
    print("🎨 GENERATING CROSS-MAP COMPARISON")
    print("="*80)
    print()
    
    # Find common agent counts across all maps
    agent_counts_sets = [set(data[map_name].keys()) for map_name in data.keys()]
    common_agent_counts = sorted(set.intersection(*agent_counts_sets)) if agent_counts_sets else []
    
    if common_agent_counts:
        for agent_count in common_agent_counts[:3]:  # Plot up to 3 common agent counts
            print(f"Creating cross-map comparison for {agent_count} agents...")
            save_path = f'{output_dir}/cross_map_comparison_{agent_count}agents.png'
            plot_cross_map_comparison(data, agent_count=agent_count, save_path=save_path)
            print()
    else:
        print("ℹ️  No common agent counts across all maps for cross-comparison")
        print()
    
    # Final summary
    print("="*80)
    print("✨ ALL PLOTS GENERATED SUCCESSFULLY!")
    print("="*80)
    print(f"📁 All plots saved to: {output_dir}/")
    print(f"📊 Total maps processed: {len(data)}")
    print(f"🎨 Total plots created: {len(data) + len(common_agent_counts[:3])}")
    print("="*80)


if __name__ == "__main__":
    # Run in script mode
    main()
else:
    # Running in Jupyter - parse data immediately
    print("🔄 Loading data for interactive use...")
    data = parse_progress_logs()
    if data:
        print(f"✅ Data loaded! Found {len(data)} maps")
        print("\nTo generate all plots, run: main()")
        print("To plot a specific map, run: plot_single_map_comprehensive('map_name', data['map_name'])")
        print("To see available maps, run: print(list(data.keys()))")
    else:
        print("❌ No data found!")
```
